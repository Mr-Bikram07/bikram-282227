```python
n = int(input("Enter a number: "))

count = 0

for i in range(1, n + 1):
    if n % i == 0:
        count = count + 1

if count == 2:
    print("Prime number")
else:
    print("Not a prime number")
```

```python
n = int(input("Enter an integer: "))
n = abs(n)
#MAKE THE NUMBER POSSITIVE
sum = 0
while n > 0:
    digit = n % 10
    sum+= digit
    n //= 10
print("Sum of digits:",sum)
```

```python
n = int(input("Enter an integer: "))
reverse = 0
while n > 0:
    digit = n % 10
    reverse = reverse * 10 + digit
    n = n // 10
print("Reversed number:", reverse)
```
